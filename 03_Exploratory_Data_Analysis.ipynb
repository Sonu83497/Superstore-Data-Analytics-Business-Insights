# Sales Analysis -

## What is the Total Sales?

### Observation

### Business Insight

### Recommendation

## What is the Average Sales per Order?

```python
average_sales = df["sales"].mean()

print(f"Average Sales per Order : ${average_sales:.2f}")
```

### Observation
The average sales value per transaction is approximately **$229.86**.

### Business Insight
Most customer orders are relatively small compared to the highest-value transactions, indicating that a few large orders contribute significantly to total revenue.

### Recommendation
Focus on increasing the average order value through product bundles, cross-selling, and upselling strategies.

## What is the Median Sales?

```python
median_sales = df["sales"].median()
print(f"Median Sales : ${median_sales:.2f}")
```

### Observation

The median sales value represents the middle transaction amount in the dataset. It is lower than the average sales, indicating that most orders are of relatively smaller value while a few high-value orders increase the overall average.

### Business Insight

The noticeable difference between the median and average sales suggests a right-skewed sales distribution. A limited number of high-value transactions contribute significantly to the company's total revenue.

### Recommendation

Increase the number of medium- and high-value orders through product bundling, cross-selling, and targeted promotional campaigns to improve overall sales performance.

## What are the Maximum and Minimum Sales?

```python
print("Maximum Sales :", df["sales"].max())
print("Minimum Sales :", df["sales"].min())
```

### Observation

The dataset contains both the highest and lowest sales transactions, showing a wide variation in order values across customers and products.

### Business Insight

The large gap between the maximum and minimum sales indicates that customer purchasing behavior varies significantly. High-value orders have a strong influence on total revenue.

### Recommendation

Identify the products, customers, and regions responsible for high-value orders and develop strategies to increase similar transactions while also improving low-value order performance.

### Business Insight

### Recommendation

Improve the average order value by implementing product recommendations, bundle offers, and customer loyalty programs. At the same time, continue nurturing high-value customers to sustain revenue growth.

## What is the Overall Profit Margin?

```python
total_sales = df['sales'].sum()
total_profit = df['profit'].sum()
profit_margin = (total_profit / total_sales) * 100
print("Overall Profit Margin :", round(profit_margin,2),"%")
```

## Do 20% of Products Generate 80% of Sales? (Pareto Analysis)

```python
product_sales = df.groupby('product_name')['sales'].sum()
product_sales = product_sales.sort_values(ascending=False)
pareto = product_sales.cumsum() / product_sales.sum() * 100
plt.figure(figsize=(12,5))
pareto.plot()
plt.axhline(80,color='red',linestyle='--')
plt.title("Pareto Analysis")
plt.ylabel("Cumulative Sales %")
plt.show()
```
